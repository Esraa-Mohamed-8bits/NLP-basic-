# Cosine Similarity Between Two Images
We load two images, turn them into gray numpy arrays, and check how similar they are using cosine similarity.

```python
# import what we need
from PIL import Image
import numpy as np
```

```python
# put the paths of your two images here
image_path_1 = "download.jfif"
image_path_2 = "download (1).jfif"
```

```python
def load_image_as_gray_array(path, size=(128, 128)):
    # open the image, make it grayscale, and resize so both images match in shape
    img = Image.open(path).convert("L")
    img = img.resize(size)
    return np.array(img)
```

```python
arr1 = load_image_as_gray_array(image_path_1)
arr2 = load_image_as_gray_array(image_path_2)

print("Image 1 shape:", arr1.shape)
print("Image 2 shape:", arr2.shape)
```

```python
def cosine_similarity(a, b):
    # flatten 2D arrays into 1D vectors first
    a = a.flatten()
    b = b.flatten()
    # cosine similarity = dot product / (length of a * length of b)
    dot_product = np.dot(a, b)
    norm_a = np.linalg.norm(a)
    norm_b = np.linalg.norm(b)
    return dot_product / (norm_a * norm_b)
```

```python
similarity = cosine_similarity(arr1, arr2)
print(f"Cosine similarity: {similarity:.4f}")
```
