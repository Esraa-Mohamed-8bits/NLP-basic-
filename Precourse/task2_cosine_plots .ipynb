# Two Cosine Plots
Plot 1: a damped cosine wave (amplitude shrinks over time).
Plot 2: a linear chirp (frequency increases from 3 Hz to 10 Hz).

```python
# import what we need
import numpy as np
import matplotlib.pyplot as plt
```

## Plot 1: Damped Cosine

```python
# time axis
t = np.linspace(0, 5, 1000)

# damping factor shrinks the wave as time goes on
damping = np.exp(-t)
frequency = 3  # Hz

damped_cosine = damping * np.cos(2 * np.pi * frequency * t)

plt.figure(figsize=(8, 4))
plt.plot(t, damped_cosine)
plt.title("Damped Cosine")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

## Plot 2: Linear Chirp (frequency goes from 3 Hz to 10 Hz)

```python
# same time axis
t2 = np.linspace(0, 5, 1000)

f0 = 3    # starting frequency
f1 = 10   # ending frequency
duration = t2[-1]

# frequency increases linearly with time, so we need the chirp rate (k)
k = (f1 - f0) / duration

# phase is the integral of frequency over time: f0*t + 0.5*k*t^2
phase = 2 * np.pi * (f0 * t2 + 0.5 * k * t2**2)
linear_chirp = np.cos(phase)

plt.figure(figsize=(8, 4))
plt.plot(t2, linear_chirp)
plt.title("Linear Chirp (f = 3 Hz to 10 Hz)")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```
